Лабораторная работа №3. Вариант 5. Метод множителей Лагранжа.

```python
import sympy as sp

x, y = sp.symbols("x y", real=True)
lam = sp.symbols("lambda", real=True)
EPS = 1e-8

f = x**2 + y**2
h = x + 2*y - 4
L = f + lam*h
J = sp.Matrix([[sp.diff(h, x), sp.diff(h, y)]])

print("Целевая функция:", f)
print("Ограничение:", sp.Eq(h, 0))
print("Функция Лагранжа:", L)
print("Матрица Якоби:", J)
print("Ранг:", J.rank())
print("Особых точек нет: ранг равен 1 везде")
```

```python
equations = [sp.diff(L, x), sp.diff(L, y), h]
print("Уравнения стационарности и ограничение:")
for eq in equations:
    print(sp.Eq(eq, 0))

solutions = sp.solve(equations, [x, y, lam], dict=True)
candidates = []
for sol in solutions:
    if not all(sol[v].is_real is True for v in [x, y, lam]):
        continue
    point = {x: sol[x], y: sol[y]}
    residual = sp.simplify(h.subs(point))
    value = sp.simplify(f.subs(point))
    allowed = abs(float(residual)) <= EPS
    print("\nКандидат:", (sol[x], sol[y]))
    print("Множитель:", sol[lam])
    print("Значение функции:", value, "=", float(value))
    print("Невязка ограничения:", residual)
    print("Допустимость:", "да" if allowed else "нет")
    if allowed:
        candidates.append((point, value))

best = None
for point, value in candidates:
    if best is None or value < best[1]:
        best = (point, value)
```

```python
# Проверка через одну переменную
t = sp.symbols("t", real=True)
x_from_h = sp.solve(h, x)[0]
g = sp.expand(f.subs(x, x_from_h).subs(y, t))
second = sp.diff(g, t, 2)
limit_plus = sp.limit(g, t, sp.oo)
limit_minus = sp.limit(g, t, -sp.oo)

print("\nПосле подстановки x =", x_from_h, ", y = t:")
print("g(t) =", g)
print("Вторая производная:", second)
print("Пределы при t -> +inf и t -> -inf:")
print(limit_plus, limit_minus)

if best is not None and second > 0:
    print("\nГлобальный минимум:", best[0])
    print("Значение минимума:", best[1])
    print("g(t) - f_min =", sp.factor(g - best[1]))
    print("Единственный минимум: g строго выпукла")
if limit_plus == sp.oo or limit_minus == sp.oo:
    print("Глобального максимума нет: g не ограничена сверху")
```
